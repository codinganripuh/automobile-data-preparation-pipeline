```python
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"],
    format= "mixed",
    dayfirst= True,
    errors= "coerce"
)
```

```python
df["transaction_date"].isnull().sum()
```

```python
df["transaction_date"] = df["transaction_date"].fillna(
    df["transaction_date"].median()
)
```

```python
df["transaction_date"].isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df["stroke"] = df["stroke"].fillna(df["stroke"].median())
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
df["price"] = df["price"].fillna(df["price"].median())
```

```python
df[["stroke", "horsepower", "price"]].isnull().sum()
```

```python
for col in ["make", "num-of-doors", "horsepower-binned"]:
    df[col] = df[col].str.strip().str.lower()
```

```python
df["make"] = df["make"].fillna(df["make"].mode()[0])
df["num-of-doors"] = df["num-of-doors"].fillna(df["num-of-doors"].mode()[0])
df["horsepower-binned"] = df["horsepower-binned"].fillna(df["horsepower-binned"].mode()[0])
```

```python
for col in df.select_dtypes(include="string").columns:
    df[col] = df[col].str.strip().str.lower()
```

```python
df.isnull().sum()
```

```python
df_processed = df.copy()
df_processed.head()
```

```python
numeric_cols = df_processed.select_dtypes(include="number").columns
```

```python
print(numeric_cols.tolist())
```

```python
col_min = df_processed[numeric_cols].min()
col_max = df_processed[numeric_cols].max()

denominator = (col_max - col_min).replace(0, 1)

df_processed[numeric_cols] = (
    df_processed[numeric_cols] - col_min
) / denominator
```

```python
df_processed[numeric_cols].describe()
```

```python
df_processed[numeric_cols].head()
```

```python
categorical_cols = df_processed.select_dtypes(
    include= ["object", "string", "category"]
).columns

print(categorical_cols.tolist())
```

```python
df_processed = pd.get_dummies(
    df_processed,
    columns=categorical_cols,
    dtype=int
)
```
